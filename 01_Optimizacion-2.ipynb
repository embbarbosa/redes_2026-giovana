```
import numpy as np
import matplotlib.pyplot as plt
```

```
alfa = 0.1
iteraciones = 15
VecX = np.zeros((iteraciones+1,1))
VecY = np.zeros((iteraciones+1,1))
VecDerivada = np.zeros((iteraciones+1,1))
VecX[0] = 2
for k in range(iteraciones+1):
    x = VecX[k]
    VecY[k] = (x-4)**2
    VecDerivada[k] = 2*(x-4)
    if k < iteraciones:
        VecX[k+1] = VecX[k]-alfa*VecDerivada[k]
x = np.linspace(-3,10,500)
plt.plot(x,(x-4)**2)
plt.plot(VecX,VecY,'o--',label='GD')
plt.grid()
plt.legend()
plt.show()
print(VecX[-1],VecY[-1])
```

```
alfa = 0.01
iteraciones = 20
VecX = np.zeros((iteraciones+1,1))
VecY = np.zeros((iteraciones+1,1))
VecDerivada = np.zeros((iteraciones+1,1))
VecX[0] = 1
for k in range(iteraciones+1):
    x = VecX[k]
    VecY[k] = (x-3)**2+7+20*np.sin(5*(x+np.pi))
    VecDerivada[k] = 2*(x-3)+100*np.cos(5*(x+np.pi))
    if k < iteraciones:
        VecX[k+1] = VecX[k]-alfa*VecDerivada[k]
x = np.linspace(-3,10,500)
plt.plot(x,(x-3)**2+7+20*np.sin(5*(x+np.pi)))
plt.plot(VecX,VecY,'o--',label='GD')
plt.grid()
plt.legend()
plt.show()
print(VecX[-1],VecY[-1])
```
