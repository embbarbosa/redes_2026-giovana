```
#Importar librerias y generar datos

import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
x1 = np.reshape(np.random.randn(100),(100,1))
x2 = np.reshape(np.random.randn(100),(100,1))
z = x1+x2
# z = -x1**2+x2
# z = x1**2+x2**2
T = (z>np.median(z)).astype(int)
plt.scatter(x1,x2,c=T,cmap='cool')
plt.grid()
plt.show()
```

```
def FunSigmoide(z):
    sigmoide = 1/(1+np.exp(-z))
    return sigmoide

def FunPolinomio(x1,x2,grado,W):
    m = x1.size
    xa = np.ones((m,grado*2+1))
    for k in range(grado):
        xa[:,[2*k+1]] = x1**(k+1)
        xa[:,[2*k+2]] = x2**(k+1)
    z = np.matmul(xa,W)
    return z

def FunCosto(Preds,T):
    m = T.size
    costo = -(1/m)*np.sum(T*np.log(Preds+1e-12)+(1-T)*np.log(1-Preds+1e-12))
    return costo

def FunDerivadas(Preds,T,x1,x2,grado):
    m = T.size
    xa = np.ones((m,grado*2+1))
    for k in range(grado):
        xa[:,[2*k+1]] = x1**(k+1)
        xa[:,[2*k+2]] = x2**(k+1)
    derivadas = np.zeros((grado*2+1,1))
    for k in range(grado*2+1):
        derivadas[k] = (1/m)*np.sum((Preds-T)*xa[:,[k]])
    return derivadas
```

```
grado = 1
alfa = 0.1
iteraciones = 100
W = np.zeros((grado*2+1,1))
Costo = np.zeros((iteraciones+1,1))
z = FunPolinomio(x1,x2,grado,W)
Preds = FunSigmoide(z)
Costo[0] = FunCosto(Preds,T)
for k in range(iteraciones):
    deriv = FunDerivadas(Preds,T,x1,x2,grado)
    W = W-alfa*deriv
    z = FunPolinomio(x1,x2,grado,W)
    Preds = FunSigmoide(z)
    Costo[k+1] = FunCosto(Preds,T)
print('Accuracy:',np.sum(np.round(Preds)==T)*100/T.size)
plt.plot(Costo)
plt.grid()
plt.show()
```

```
x1_prueba = np.linspace(np.min(x1)-.5,np.max(x1)+.5,100)
x2_prueba = np.linspace(np.min(x2)-.5,np.max(x2)+.5,100)
MatrizPred = np.zeros((x2_prueba.size,x1_prueba.size))
for i in range(x1_prueba.size):
    for j in range(x2_prueba.size):
        z = FunPolinomio(np.array([[x1_prueba[i]]]),np.array([[x2_prueba[j]]]),grado,W)
        MatrizPred[j,i] = FunSigmoide(z)[0,0]
plt.pcolormesh(x1_prueba,x2_prueba,MatrizPred,cmap='RdBu',shading='auto',vmin=0,vmax=1)
plt.colorbar()
plt.scatter(x1,x2,c=T,cmap='cool',edgecolor='k')
plt.show()
```
